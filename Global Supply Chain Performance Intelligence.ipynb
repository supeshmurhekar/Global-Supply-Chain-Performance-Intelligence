# Clean Shipment Table

```python
shipment['Dispatch_Date'] = pd.to_datetime(
    shipment['Dispatch_Date'],
    dayfirst=True
)

shipment['Delivery_Date'] = pd.to_datetime(
    shipment['Delivery_Date']
)
```

```python
shipment.isna().sum()
```

```python
shipment['Actual_Days'] = shipment['Actual_Days'].fillna(
shipment['SLA_Days']
)
```

```python
# Fix negative shipping cost
shipment.loc[
    shipment['Total_Shipping_Cost'] < 0,
    'Total_Shipping_Cost'
] = abs(shipment['Total_Shipping_Cost'])
```

```python
# Recalculate Profit
shipment['Profit'] = (
    shipment['Revenue'] -shipment['Total_Shipping_Cost']
)
```

```python
shipment
```

# 1. Transit Time
actual shipment duration.

```python
shipment["Transit_Time"] = (
    shipment['Delivery_Date']-shipment['Dispatch_Date']
)
```

# 2. Delay Days
How many days the shipment/order is late

```python
shipment['Delay_Days']=(
    shipment['Actual_Days']-shipment['SLA_Days']
)
```

# 3.Delay Flag
Whether it is late or not (binary indicator)
0 ---> on time 
1 ---> Delayed

```python
shipment['Delay_Flag'] = (
    shipment['Actual_Days'].fillna(0) >
    shipment['SLA_Days'].fillna(0)
).astype(int)
```

# 4.Cost per KM

```python
shipment['Cost_per_KM'] = (
   round( shipment['Total_Shipping_Cost']/shipment['Distance_KM'] ,2)
)
```

```python
shipment
```

# 5.Revenue per KM

```python
shipment['Revenue_per_KM'] =(
   round( shipment['Revenue']/shipment['Distance_KM'])
)
```

# 6.. Profit Margin %

```python
shipment['Profit_Margin_%'] = (
  round(  shipment['Profit'] / shipment['Revenue'] * 100)
)
```

# 7. Inventory Health Flag  
it is a binary business indicator that tells whether current inventory is in a healthy state based on predefined supply thresholds.
0 ---> healthy 
1 ----> unhealthy 

```python
inventory['Stock_Gap'] = (
    inventory['Closing_Stock']
    - (inventory['Opening_Stock']
       + inventory['Inbound_Qty']
       - inventory['Outbound_Qty'])
)
```

 This returns Boolean values:
 True --> mismatch exists
 False --> perfectly matched
 This converts:
 True --> 1
 False --> 0

```python
inventory['Stock_Mismatch_Flag'] = (
    inventory['Stock_Gap'] != 0
).astype(int)
```

# 8.Purchase Lead Time
How many days it takes to receive goods after placing a purchase order.  OR It is the waiting time between ordering and receiving.

```python
purchase['Lead_Time_Actual'] = (
    purchase['Actual_Delivery_Date']
    - purchase['Order_Date']
).dt.days
```

# 10.Shipment Cost Deviation %
How much the actual shipment cost is higher or lower than the planned (expected) cost


```python
avg_cost_by_mode = shipment.groupby('Transport_Mode')['Total_Shipping_Cost'].transform('mean')

shipment['Cost_Deviation_%'] = (
    (shipment['Total_Shipping_Cost'] - avg_cost_by_mode)
    / avg_cost_by_mode * 100
)
```

# 11. Warehouse Load Utilization Proxy
An estimated indicator of how much warehouse capacity is being used.
Because many companies don’t have exact capacity data, they create a proxy (approximate measure).

```python
inventory['Throughput'] = (
    inventory['Inbound_Qty'] + inventory['Outbound_Qty']
)

inventory['Warehouse_Load_Index'] = (
    inventory['Throughput'] /
    inventory.groupby('Warehouse_ID')['Throughput'].transform('mean')
)
```

# 12. Order Fulfillment Speed Bucket
A category that groups orders based on how fast they were fulfilled.

# Transit Time 
How many days a shipment takes to travel from dispatch to delivery.  or Shipment ko warehouse se customer tak pahunchne me kitna time laga?

```python
# shipment.drop(columns=['Fulfillment_Speed'],axis=1,inplace=True)
```

```python
shipment['Transit_Time']=(shipment['Delivery_Date']-shipment['Dispatch_Date'])
```

```python
shipment['Transit_Time_Days'] = shipment['Transit_Time'].dt.days
```

```python
shipment['Fulfillment_Speed'] = pd.cut(
    shipment['Transit_Time_Days'],
    bins=[0, 3, 6, 10, np.inf],
    labels=['Express Fast', 'Standard', 'Slow', 'Very Slow'],
    include_lowest=True
)
```
